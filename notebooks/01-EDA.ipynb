```python
numeric_cols = data.select_dtypes(include="number").columns
# print(numeric_cols)
corr = data[numeric_cols].corr()
plt.figure(figsize=(16, 12))

plt.imshow(corr, aspect="auto")
plt.colorbar()
plt.xticks(
  range(len(numeric_cols)),
  numeric_cols,
  rotation=90
)
plt.yticks(
    range(len(numeric_cols)),
    numeric_cols
)
plt.title("Correlation Matrix")
plt.show()
```

Here we look specifically at which numerical statistics have the strongest positive relationship with goals. For example, `Shots` has a correlation of 0.778742, meaning players with more shots tend to have higher goal totals in this dataset.

```python
goal_corr = (
    corr["Goals"]
    .drop("Goals")
    .sort_values(ascending=False)
)

goal_corr.head(10)
```

Percentiles show how playing time is distributed among players. For example, the 50th percentile is the median, meaning half of the players played fewer minutes than this value and half played more.

```python
data["Minutes"].quantile([0, 0.05, 0.10, 0.25, 0.50, 0.75, 0.90, 0.95, 1])
```

```python
thresholds = [450, 900, 1350]

for threshold in thresholds:
  count = (data["Minutes"] < threshold).sum()
  percentage = count / len(data) * 100

  print(
      f"< {threshold} minutes: "
      f"{count} players ({percentage:.2f}%)"
  )
```

This shows which statistics have the strongest positive relationship with total playing minutes. A high positive correlation means players who play more minutes tend to accumulate more of that statistic, which can reveal which features are strongly affected by playing time.

```python
minutes_corr = (
    corr["Minutes"]
    .drop("Minutes")
    .sort_values(ascending=False)
)

minutes_corr
```

This extracts pairs of numerical features with strong positive correlations so that we can identify statistics that tend to move together. Highly correlated features may contain overlapping information.

```python
corr_pairs = (
    corr.where(
        ~__import__("numpy").tril(
            __import__("numpy").ones(corr.shape)
        ).astype(bool)
    )
    .stack()
    .sort_values(ascending=False)
)

corr_pairs.head(20)
```

This shows pairs of features with the strongest negative relationships. Negative correlation means that players with higher values for one statistic tend to have lower values for another, which may reveal differences between player roles or playing styles.
